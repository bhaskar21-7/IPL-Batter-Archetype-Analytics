```python
missing = df.isnull().sum()
missing = missing[missing>0]
missing.sort_values(ascending=False)
```

```python
missing_pct = (df.isnull().sum())/(len(df))

missing_pct = missing_pct[missing_pct>0]

missing_pct.sort_values(ascending=False)
```
